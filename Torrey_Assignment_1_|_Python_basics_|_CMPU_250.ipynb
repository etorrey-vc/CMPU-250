# Control structures

As with many languages, Python has control structures to dictate the flow of code execution, including the main two: branching (conditionals) and iteration (loops).

## Conditionals

As their name indicates, conditionals are a way to execute code depending on whether a condition is True or False. As in other languages, Python supports `if` and `else` but `else if` is contracted into `elif`, as the example below demonstrates.

````
num = 5

if num < 0:
  print("negative")

elif num == 0:
  print("null")

else: # num > 0
  print("positive")
````

Here `<` and `>` are the strict `less` and `greater than` operators and `==` is the equality operator (not to be confused with `=`, the assignment operator). The operators `<=` and `>=` can be used for less (resp. greater) than or equal comparisons.

Contrary to other languages, blocks of code are delimited using indentation. Here, we use 2-space indentation but many programmers also use 4-space indentation. Any one is fine as long as you are consistent throughout your code.

## Loops

Loops are a way to execute a block of code multiple times.

There are two main types of loops: **while loops** and **for loops**.

For loops execute the indented code a set number of times. Each iteration of the loop starts by assigning the iterator (e.g., `i` below) to the next value in an iterable (e.g., `range(len(my_list))` below). Then, the indented code executes. Once the loop reaches the end of the iterable, the loop exits and continues executing any code below it in the cell.

** Note: In Python, iterators in for-loops have global scope. Below, once the loop exits `i` will be assigned to the last element in `range(len(my_list))`.

````
for i in range(len(my_list)):
  print(my_list[i])
````

If the goal is simply to iterate over a list, we can do so directly as follows

** Note: It's good practice to name the iterator `i` if you're iterating through an index of a list and name it `element` or something more descriptive if you're looping over elements in a sequence.

````
for element in my_list:
  print(element)
````

While loops execute the indented code as long as the condition in the first line of the while statement is true. For example, below, once `i` is greater than or equal to `len(my_list)`, the loop will exit and continue on executing anything else in the cell.

````
i = 0

while i < len(my_list):
  print(my_list[i])
  i += 1 # equivalent to i = i + 1
````

````
def connect_to_next_port(self, minimum: int) -> int:
    """Connects to the next available port.

    Args:
      minimum: A port value greater or equal to 1024.

    Returns:
      The new minimum port.

    Raises:
      ValueError: If port is less than minimum
      ConnectionError: If no available port is found.
    """

    if minimum < 1024:
      raise ValueError(f'Min. port must be at least 1024, not {minimum}.')

    port = self._find_next_open_port(minimum)

    if port is None:
      raise ConnectionError(
          f'Could not connect to service on port {minimum} or higher.')

    assert port >= minimum, (
        f'Unexpected port {port} when minimum was {minimum}.')

    return port
````

# Classes & objects

We can use Python to do object-oriented programming. We can define **classes** with **attributes** and **methods**. A class definition makes a constructor that we can use to instantiate new objects of that class type, as well as methods that we can call with those objects.

For example, I can define a class that makes Pokemon objects with **attributes** like hp or species and **methods** like if the Pokemon gets attacked.

````
class Pokemon:

  # constructor
  def __init__(self, name : str, types : list[str], hp : int, fainted : bool):
    # attributes
    self.name = name
    self.types = types
    self.hp = hp
    self.fainted = fainted

  # methods
  def attacked(self, hp_decrease_amount):
    '''Takes in the amount of hp that pokemon is attacked for, decreases hp
    and sets fainted=True if hp is 0'''

    self.hp -= hp_decrease_amount

    # set fainted=True
    if self.hp <= 0:
      self.hp = 0
      self.fainted = True
````

````
# constructor example
pokemon1 = Pokemon(name = 'Clefairy', types = ["fairy"], hp = 70, fainted = False)

# access attributes
print(pokemon1.hp)
print(pokemon1.fainted)
````

````
# method example
pokemon1.attacked(100)

print(pokemon1.hp)
print(pokemon1.fainted)
````

We often don't need to define classes in data science applications, but they'll be useful sometimes.

Use the `time` module to time how long (in seconds) your `sieve` code takes to execute. Read the documentation to figure out how to use this module: https://docs.python.org/3/library/time.html

Write a function named `execution_time` that takes in a function `fun` and a number `n` and returns the number of seconds it takes to execute `fun(n)`.

Then, write an expression to print the number of seconds it takes to run your `sieve` code on the following list of inputs: `[1, 10, 100, 1000, 10000, 100000, 1000000]`.

As a rule of thumb, depending on how I write the `sieve` function, it runs for somewhere  between 3 to 30 seconds for `n=1000000`.

````
import time

def sieve(n : int) -> list[int]:
  primes = list(range(2,n))

  i = 0
  p = 2

  while (p <= n ** (1/2)):

    primes = list(filter(lambda x : x % p != 0 or x == p, primes))

    i += 1
    p = primes[i]

  return primes
````

````
def execution_time( fun, n : int ) -> int:
    '''that takes in a function fun and a number n and returns the number of seconds it takes to execute fun(n)'''

    start = time.time()
    fun(n)
    end = time.time()
    duration = end - start

    print(f"function on {n} took {duration} seconds")

inputs = [1, 10, 100, 1000, 10000, 100000, 1000000]

for input in inputs:
    execution_time(sieve, input)
````

**Exercise 6.** Write a function that takes in an Inventory and a grocery list and returns the total price of a grocery list.

Your function should take in a grocery list, which is a list of tuples in the format `("<portion of item name>", <quantity>)`, and output the total cost of the least expensive item in the inventory. If the item name is missing some specification, default to the cheapeast item in the inventory that matches the portion of item name given.

For example, if I input the following grocery list with the inventory created in Exercise 6:

`[('Honeycrisp apples', 2), ('whole milk (1 gal)', 1), ('cabbage',2)]`

your function should return 12.73.

````
def total_price( inventory : Inventory, grocery_list : list[tuple[str, float]]) -> float :
    '''takes a list of tuples in the format ("<portion of item name>", <quantity>), and output the total cost of the grocery list'''
    inventory_name_dict = inventory.name_map()
    total = 0.0

    for item in grocery_list:

        #if there is an item that matches the str, add the price * quantity to total
        if item[0] in inventory_name_dict:
            total += (item[1] * inventory_name_dict[item[0]])

        #if no item matches, add the closest cheapest item
        else:
            cheapest = None
            for inventory_item in inventory.items:
                if item[0] in inventory_item.name:
                    if cheapest is None or inventory_item.price < cheapest.price:
                        cheapest = inventory_item

            total += (item[1] * cheapest.price)

    return total

total_price(inventory1, [('Honeycrisp apples', 2), ('whole milk (1 gal)', 1), ('cabbage',2)])
````
